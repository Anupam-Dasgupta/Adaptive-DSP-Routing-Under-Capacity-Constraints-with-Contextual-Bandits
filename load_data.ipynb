```python
from datasets import load_dataset

dataset = load_dataset("EvgeniaKyriazi/ctr-prediction-dataset")
dataset
```

```python
from pathlib import Path

output_dir = Path("data/raw")
output_dir.mkdir(parents=True, exist_ok=True)

paths = {
    "train": output_dir / "avazu_train.parquet",
    "test": output_dir / "avazu_test.parquet",
}
for split, path in paths.items():
    dataset[split].to_pandas().to_parquet(path, index=False)
    print(f"Saved {split}: {path} ({len(dataset[split]):,} rows)")
```
