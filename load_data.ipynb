```python
from datasets import load_dataset

ds = load_dataset("EvgeniaKyriazi/ctr-prediction-dataset")

print(ds)
print(ds["train"].column_names)
```

```python
from pathlib import Path

# Create output directory
output_dir = Path("data")
output_dir.mkdir(parents=True, exist_ok=True)

# Convert and save train split
train_df = ds["train"].to_pandas()
train_path = output_dir / "avazu_train.parquet"
train_df.to_parquet(train_path, index=False)

# Convert and save test split
test_df = ds["test"].to_pandas()
test_path = output_dir / "avazu_test.parquet"
test_df.to_parquet(test_path, index=False)

print(f"Saved train split to: {train_path}")
print(f"Saved test split to:  {test_path}")
print(f"Train shape: {train_df.shape}")
print(f"Test shape:  {test_df.shape}")
```
